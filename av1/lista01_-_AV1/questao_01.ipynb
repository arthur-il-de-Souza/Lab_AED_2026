### 1. O maior número ímpar

```python
# lista
v = [9, 42, 21, 14, 28, 3, 19, 32, 46, 6]
```

```python
maior = 0
for i in range(0, len(v)-1, 1):
    if v[i+1] > v[i] and v[i+1]%2==1:
        maior = v[i+1] 
print(f"{v}")
if maior != 0: print(f"Maior elemento ímpar da lista: {maior}")
else: print(f"Não existe elementos impares na lista!!!")
```
