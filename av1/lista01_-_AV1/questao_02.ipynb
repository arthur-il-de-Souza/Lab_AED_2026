### 2. O segundo maior ímpar

```python
# Lista
v = [9, 42, 21, 14, 28, 3, 19, 32, 46, 6]
aux = [0]*2
```

```python
for i in range(0, len(v)-1, 1):
    if v[i]%2==1:
        if v[i] > aux[1]:
            aux[1] = v[i]
        elif v[i] > aux[0]:
            aux[0] = v[i]
if(aux[0] != 0): print(f"O segundo maior número ímpar na lista é: {aux[0]}")
else: print(f"Não existe número ímpar na lista!!!")
```
