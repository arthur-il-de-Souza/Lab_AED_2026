### 4. Ímpar-Ímpar

```python
# Lista
v = [9, 2, 7, 7, 2, 2, 1, 7, 7, 9]
```

```python
for i in range(0, len(v)-1, 1):
    cont = 0
    if v[i]%2==0: continue
    for j in range(0, len(v), 1):
        if v[j] == v[i]: cont+=1    
    if(cont%2 == 1):
        print(f"O número {v[i]} aparece {cont} vezes!!!")
        break

```
