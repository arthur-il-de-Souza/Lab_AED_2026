### 6. Os dois elementos mais próximos

```python
# Lista
v = [9, 42, 21, 14, 25, 3, 19, 33, 45, 6]
aux = [0]*2
```

```python
dist = abs(v[0] - v[1])
aux[0], aux[1] = v[0], v[1]

for i in range(0, len(v)-1, 1):
    for j in range(i+1, len(v), 1):
        if dist > abs(v[i] - v[j]):
            dist = abs(v[i] - v[j])
            aux[0], aux[1] = v[i], v[j]

print(f"A menor distância entre dis valores é {dist}, valores: {aux[0]} e { aux[1]}")
```
