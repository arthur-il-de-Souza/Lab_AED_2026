### 3. O mais próximo da média

```python
# Lista
v = [5, 3, 1, 10, 2, 13, 9, 12, 4, 7]
```

```python
media = 0
for i in v:
    media += i
media = media/len(v)

dist = abs(v[0] - media)
num = v[0]

for i in range(1, len(v), 1):
    if abs(v[i]-media) < dist:
        dist = abs(v[i]-media)
        num = v[i]

print(f"O número que mais se aproxima da média {media} é {num}")
```
