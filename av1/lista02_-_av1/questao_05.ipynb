### 5. k repeticões

```python
# Lista
v = [7, 1, 9, 1, 7, 3, 9, 2, 1, 6, 8, 3]
```

```python
k = 3

for i in range(0, len(v), 1):
    cont = 0
    for j in range(0, len(v), 1):
        if v[i] == v[j]: cont +=1
    if cont == k:
        print(f"O número {v[i]} repete {cont} vezes!!!")
        break
```
