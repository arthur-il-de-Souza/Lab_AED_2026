### 2. O k-ésimo maior elemento

```python
# Lista
v = [9, 42, 21, 14, 25, 3, 19, 33, 45, 6]
```

```python
k = len(v)

for i in range(0, k, 1):
    for j in range(0, len(v)-(i+1), 1):
        if(v[j] > v[len(v)-(i+1)]):
            aux = v[len(v)-(i+1)]
            v[len(v)-(i+1)] = v[j]
            v[j] = aux
print(v)
```
