#### 4. Elemento isolado

```python
# Lista
v = [17, 2, 8, 1, 7, 13, 9, 12, 4, 16]
```

```python
# Ordenando a lista para facilitar a busca pelo elemento isolado
for i in range(0, len(v), 1):
    for j in range(0, len(v)-(i+1), 1):
        if(v[j] > v[len(v)-(i+1)]):
            aux = v[len(v)-(i+1)]
            v[len(v)-(i+1)] = v[j]
            v[j] = aux

for i in range(1, len(v)-1,1):
    if v[i-1] != v[i]-1 or v[i+1] != v[i]+1:
        print(f"O número {v[i]} é um elemento isolado!!!")
        # break
```
