### 1. Imagine que nós aplicamos o procedimento Partição() sobre a lista L 
<p>
    E imagine que depois nós ordenamos os dois lados da partição usando o algoritmo da bolha<br>
    Claramente, esse procedimento ordena a lista completamente.
</p>

#### a) Analise o tempo de execução desse algoritmo no melhor caso.

No melhor caso, onde as duas metades foram organizadas de modo que qualquer $x \in l_2$, onde $ l_2 $ é um subconjunto de $ L $, tal que,  para qualquer elemento de $ l_1 $, também um subconjunto de $ L $, vale a seguinte relação: 

$
    \forall y \in l_{1}, \forall x \in l_{2}|y_{i} \leq x_{i}, 0 \leq i \leq n
$

Dada esse condição o algoritmo rodará em $\mathcal{O}(n)$, pois percorrerá a lista sem executar nenhuma mudança.

#### b) Analise o tempo de execução desse algoritmo no pior caso.
No pior dos casos o tempo do algoritmo é $ \mathcal{O}(\frac{n^2}{k}) $, onde $ k $ é a razão entre a menor quantidade de elementos da menor porção ordenada e o número total de elementos.
