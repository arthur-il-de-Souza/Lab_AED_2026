### 5. Vetores esparsos

```python
class NoDuplo:
    def __init__(self, valor, indice=None):
        if isinstance(valor, NoDuplo):
            self.valor = valor.valor
            self.indice = valor.indice
        else:
            self.valor = valor
            self.indice = indice
        self.anterior = None
        self.proximo = None
```

```python
class ListaDuplamenteEncadeada:
    def __init__(self):
        self.inicio = None
        self.fim = None

    # Métodos CRUD
    def criar(self, valor, indice=None):
        novo = valor if isinstance(valor, NoDuplo) else NoDuplo(valor, indice)
        if not self.inicio:
            self.inicio = novo
            self.fim = novo
            return
        self.fim.proximo = novo
        novo.anterior = self.fim
        self.fim = novo

    def ler(self):
        elementos = []
        atual = self.inicio
        while atual:
            elementos.append((atual.valor, atual.indice))
            atual = atual.proximo
        return elementos

    def atualizar(self, indice, valor_novo):
        atual = self.inicio
        while atual:
            if atual.indice == indice:
                atual.valor = valor_novo
                return True
            atual = atual.proximo
        return False

    def deletar(self, indice):
        atual = self.inicio
        while atual:
            if atual.indice == indice:
                if atual.anterior:
                    atual.anterior.proximo = atual.proximo
                else:
                    self.inicio = atual.proximo

                if atual.proximo:
                    atual.proximo.anterior = atual.anterior
                else:
                    self.fim = atual.anterior
                return True
            atual = atual.proximo
        return False

    # Solução da Questão (a)
    @classmethod
    def construir_de_vetor(cls, V, index_base=1):
        lista = cls()
        for i, valor in enumerate(V):
            if valor != 0:
                indice = i + index_base  # Considera indexação a partir de 1 como no exemplo
                lista.criar(valor, indice)
        return lista
```
