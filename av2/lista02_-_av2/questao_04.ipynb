### 4. Partição

```python
class NoDuplo:
    def __init__(self, valor):
        self.valor = valor.valor if isinstance(valor, NoDuplo) else valor
        self.anterior = None
        self.proximo = None
```

```python
class ListaDuplamenteEncadeada:
    def __init__(self):
        self.inicio = None
        self.fim = None

    # Métodos CRUD
    def criar(self, valor):
        novo = valor if isinstance(valor, NoDuplo) else NoDuplo(valor)
        if not self.inicio:
            self.inicio = novo
            self.fim = novo
            return
        self.fim.proximo = novo
        novo.anterior = self.fim
        self.fim = novo

    def ler(self):
        elementos = []
        atual = self.inicio
        while atual:
            elementos.append(atual.valor)
            atual = atual.proximo
        return elementos

    def atualizar(self, valor_antigo, valor_novo):
        atual = self.inicio
        while atual:
            if atual.valor == valor_antigo:
                atual.valor = valor_novo
                return True
            atual = atual.proximo
        return False

    def deletar(self, valor):
        atual = self.inicio
        while atual:
            if atual.valor == valor:
                if atual.anterior:
                    atual.anterior.proximo = atual.proximo
                else:
                    self.inicio = atual.proximo

                if atual.proximo:
                    atual.proximo.anterior = atual.anterior
                else:
                    self.fim = atual.anterior
                return True
            atual = atual.proximo
        return False

    # Solução da Questão (a)
    def particionar(self, k):
        if not self.inicio or self.inicio == self.fim:
            return self

        q = self.inicio
        r = self.fim

        # Executa enquanto q e r não se encontrarem nem se cruzarem
        while q and r and q != r and q.anterior != r:
            # 1. Percorre da esquerda para a direita até achar valor > k
            while q and q != r and q.anterior != r and q.valor <= k:
                q = q.proximo

            # 2. Percorre da direita para a esquerda até achar valor <= k
            while r and r != q and r.proximo != q and r.valor > k:
                r = r.anterior

            # 3. Se q e r ainda não se cruzaram, troca os elementos de posição
            if q and r and q != r and q.anterior != r and q.valor > k and r.valor <= k:
                q.valor, r.valor = r.valor, q.valor
                q = q.proximo
                r = r.anterior

        return self
```
