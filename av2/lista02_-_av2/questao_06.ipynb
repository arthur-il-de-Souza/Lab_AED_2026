### 6. Operaçãoes sobre vetores esparsos

```python
class NoDuplo:
    def __init__(self, valor, indice=None):
        if isinstance(valor, NoDuplo):
            self.valor = valor.valor
            self.indice = valor.indice
        else:
            self.valor = valor
            self.indice = indice
        self.anterior = None
        self.proximo = None
```

```python
class VetorEsparso:
    def __init__(self):
        self.inicio = None
        self.fim = None

    # Métodos CRUD
    def criar(self, valor, indice):
        novo = valor if isinstance(valor, NoDuplo) else NoDuplo(valor, indice)
        if not self.inicio:
            self.inicio = novo
            self.fim = novo
            return
        self.fim.proximo = novo
        novo.anterior = self.fim
        self.fim = novo

    def ler(self):
        elementos = []
        atual = self.inicio
        while atual:
            elementos.append((atual.valor, atual.indice))
            atual = atual.proximo
        return elementos

    def atualizar_simples(self, indice, valor_novo):
        atual = self.inicio
        while atual:
            if atual.indice == indice:
                atual.valor = valor_novo
                return True
            atual = atual.proximo
        return False

    def deletar(self, indice):
        atual = self.inicio
        while atual:
            if atual.indice == indice:
                if atual.anterior:
                    atual.anterior.proximo = atual.proximo
                else:
                    self.inicio = atual.proximo

                if atual.proximo:
                    atual.proximo.anterior = atual.anterior
                else:
                    self.fim = atual.anterior
                return True
            atual = atual.proximo
        return False

    # Solução da Questão (a1)
    def busca_por_indice(self, k):
        atual = self.inicio
        while atual:
            if atual.indice == k:
                return atual.valor
            if atual.indice > k:  # Como os índices são ordenados, interrompe antecipadamente
                break
            atual = atual.proximo
        return 0  # Elementos não armazenados na lista têm valor zero

    # Solução da Questão (a2)
    def busca_por_valor(self, x):
        atual = self.inicio
        while atual:
            if atual.valor == x:
                return atual.indice
            atual = atual.proximo
        return -1  # Retorna -1 se o valor não estiver na lista

    # Solução da Questão (a3)
    def atualizacao(self, x, k):
        atual = self.inicio
        ant = None

        # Procura a posição do índice k (ou a posição correta para inserção)
        while atual and atual.indice < k:
            ant = atual
            atual = atual.proximo

        # Caso 1: O índice k já existe na lista
        if atual and atual.indice == k:
            if x == 0:
                # Atualizar para 0 implica remover o nó para preservar a esparsidade
                if atual.anterior:
                    atual.anterior.proximo = atual.proximo
                else:
                    self.inicio = atual.proximo

                if atual.proximo:
                    atual.proximo.anterior = atual.anterior
                else:
                    self.fim = atual.anterior
            else:
                atual.valor = x
            return

        # Caso 2: O índice k não existe na lista e x != 0 (inserção ordenada)
        if x != 0:
            novo = NoDuplo(x, k)
            novo.anterior = ant
            novo.proximo = atual

            if ant:
                ant.proximo = novo
            else:
                self.inicio = novo

            if atual:
                atual.anterior = novo
            else:
                self.fim = novo
```
