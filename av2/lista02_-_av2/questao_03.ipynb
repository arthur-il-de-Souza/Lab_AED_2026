3. Varredura e ordenação

```python
class NoDuplo:

  def __init__(self, valor):
    self.valor = valor.valor if isinstance(valor, NoDuplo) else valor
    self.anterior = None
    self.proximo = None
```

```python
class ListaDuplamenteEncadeada:

  def __init__(self):
    self.inicio = None
    self.fim = None

  # Métodos CRUD
  def criar(self, valor):
    novo = valor if isinstance(valor, NoDuplo) else NoDuplo(valor)
    if not self.inicio:
      self.inicio = novo
      self.fim = novo
      return
    self.fim.proximo = novo
    novo.anterior = self.fim
    self.fim = novo

  def ler(self):
    elementos = []
    atual = self.inicio
    while atual:
      elementos.append(atual.valor)
      atual = atual.proximo
    return elementos

  def atualizar(self, valor_antigo, valor_novo):
    atual = self.inicio
    while atual:
      if atual.valor == valor_antigo:
        atual.valor = valor_novo
        return True
      atual = atual.proximo
    return False

  def deletar(self, valor):
    atual = self.inicio
    while atual:
      if atual.valor == valor:
        if atual.anterior:
          atual.anterior.proximo = atual.proximo
        else:
          self.inicio = atual.proximo

        if atual.proximo:
          atual.proximo.anterior = atual.anterior
        else:
          self.fim = atual.anterior
        return True
      atual = atual.proximo
    return False

  # Método auxiliar para trocar fisicamente dois nós adjacentes (no1 e no2)
  def _trocar_nos_adjacentes(self, no1, no2):
    if not no1 or not no2 or no1.proximo != no2:
      return

    ant = no1.anterior
    prox = no2.proximo

    if ant:
      ant.proximo = no2
    else:
      self.inicio = no2

    if prox:
      prox.anterior = no1
    else:
      self.fim = no1

    no2.anterior = ant
    no2.proximo = no1
    no1.anterior = no2
    no1.proximo = prox

  # Solução da Questão (a)
  def varredura(self):
    if not self.inicio or not self.inicio.proximo:
      return self

    atual = self.inicio
    while atual and atual.proximo:
      if atual.valor > atual.proximo.valor:
        prox_no = atual.proximo
        self._trocar_nos_adjacentes(atual, prox_no)
        # Após a troca de nós, 'atual' avança para a direita e continua sendo testado no próximo passo
      else:
        atual = atual.proximo

    return self
```
