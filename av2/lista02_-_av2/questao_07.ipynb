### 10. Operaçãoes na lista de listas

```python
class NoDuplo:
    def __init__(self, valor):
        self.valor = valor.valor if isinstance(valor, NoDuplo) else valor
        self.anterior = None
        self.proximo = None
```

```python
class ListaDuplamenteEncadeada:
    def __init__(self):
        self.inicio = None
        self.fim = None

    # Métodos CRUD da Lista Duplamente Encadeada
    def criar(self, valor):
        novo = valor if isinstance(valor, NoDuplo) else NoDuplo(valor)
        if not self.inicio:
            self.inicio = novo
            self.fim = novo
            return
        self.fim.proximo = novo
        novo.anterior = self.fim
        self.fim = novo

    def ler(self):
        elementos = []
        atual = self.inicio
        while atual:
            elementos.append(atual.valor)
            atual = atual.proximo
        return elementos

    def atualizar(self, valor_antigo, valor_novo):
        atual = self.inicio
        while atual:
            if atual.valor == valor_antigo:
                atual.valor = valor_novo
                return True
            atual = atual.proximo
        return False

    def deletar(self, valor):
        atual = self.inicio
        while atual:
            if atual.valor == valor:
                if atual.anterior:
                    atual.anterior.proximo = atual.proximo
                else:
                    self.inicio = atual.proximo

                if atual.proximo:
                    atual.proximo.anterior = atual.anterior
                else:
                    self.fim = atual.anterior
                return True
            atual = atual.proximo
        return False


class ListaDeListas:
    def __init__(self):
        self.sublistas = []  # Vetor/lista de ponteiros para as sublistas

    # Métodos CRUD da ListaDeListas
    def criar_sublista(self):
        nova_sublista = ListaDuplamenteEncadeada()
        self.sublistas.append(nova_sublista)
        return nova_sublista

    def ler_tudo(self):
        return [sub.ler() for sub in self.sublistas]

    # Solução da Questão (a1): Busca(L, x)
    def busca(self, x):
        for i, sublista in enumerate(self.sublistas):
            # Aproveita a ordenação dos intervalos das sublistas
            if sublista.fim and x <= sublista.fim.valor:
                atual = sublista.inicio
                while atual:
                    if atual.valor == x:
                        return (i, atual)  # Retorna o índice da sublista e a referência do nó
                    if atual.valor > x:
                        break
                    atual = atual.proximo
                return None
        return None

    # Solução da Questão (a2): Inserção(L, x)
    def insercao(self, x):
        if not self.sublistas:
            nova_sub = self.criar_sublista()
            nova_sub.criar(x)
            return

        # Localiza a sublista adequada baseando-se no limite superior do intervalo
        target_sublista = self.sublistas[-1]
        for sublista in self.sublistas:
            if not sublista.inicio or x <= sublista.fim.valor:
                target_sublista = sublista
                break

        # Insere x mantendo a ordem crescente na sublista selecionada
        novo = NoDuplo(x)
        if not target_sublista.inicio:
            target_sublista.inicio = novo
            target_sublista.fim = novo
            return

        atual = target_sublista.inicio
        while atual and atual.valor < x:
            atual = atual.proximo

        if not atual:  # Inserção no final da sublista
            target_sublista.fim.proximo = novo
            novo.anterior = target_sublista.fim
            target_sublista.fim = novo
        elif not atual.anterior:  # Inserção no início da sublista
            novo.proximo = target_sublista.inicio
            target_sublista.inicio.anterior = novo
            target_sublista.inicio = novo
        else:  # Inserção no meio da sublista
            ant = atual.anterior
            ant.proximo = novo
            novo.anterior = ant
            novo.proximo = atual
            atual.anterior = novo

    # Solução da Questão (a3): Remoção(L, x)
    def remocao(self, x):
        resultado_busca = self.busca(x)
        if not resultado_busca:
            return False

        idx_sublista, no_alvo = resultado_busca
        sublista = self.sublistas[idx_sublista]

        # Remoção do nó ajustando os ponteiros duplos
        if no_alvo.anterior:
            no_alvo.anterior.proximo = no_alvo.proximo
        else:
            sublista.inicio = no_alvo.proximo

        if no_alvo.proximo:
            no_alvo.proximo.anterior = no_alvo.anterior
        else:
            sublista.fim = no_alvo.anterior

        return True
```
