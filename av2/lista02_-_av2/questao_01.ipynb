### 1. O elemento central

```python
class NoDuplo:
    def __init__(self, valor):
        self.valor = valor.valor if isinstance(valor, NoDuplo) else valor
        self.anterior = None
        self.proximo = None
```

```python
class ListaDuplamenteEncadeada:
    def __init__(self):
        self.inicio = None
        self.fim = None

    # Métodos CRUD
    def criar(self, valor):
        novo = valor if isinstance(valor, NoDuplo) else NoDuplo(valor)
        if not self.inicio:
            self.inicio = novo
            self.fim = novo
            return
        self.fim.proximo = novo
        novo.anterior = self.fim
        self.fim = novo

    def ler(self):
        elementos = []
        atual = self.inicio
        while atual:
            elementos.append(atual.valor)
            atual = atual.proximo
        return elementos

    def atualizar(self, valor_antigo, valor_novo):
        atual = self.inicio
        while atual:
            if atual.valor == valor_antigo:
                atual.valor = valor_novo
                return True
            atual = atual.proximo
        return False

    def deletar(self, valor):
        atual = self.inicio
        while atual:
            if atual.valor == valor:
                if atual.anterior:
                    atual.anterior.proximo = atual.proximo
                else:
                    self.inicio = atual.proximo

                if atual.proximo:
                    atual.proximo.anterior = atual.anterior
                else:
                    self.fim = atual.anterior
                return True
            atual = atual.proximo
        return False

    # Solução da Questão (a)
    def imprimir_elemento_central(self):
        if not self.inicio:
            print("Lista vazia.")
            return None

        lento = self.inicio
        rapido = self.inicio

        while rapido.proximo and rapido.proximo.proximo:
            lento = lento.proximo
            rapido = rapido.proximo.proximo

        print(f"Elemento central: {lento.valor}")
        return lento.valor
```
