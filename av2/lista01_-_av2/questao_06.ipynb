### 6. Partição

```python
class No:
    def __init__(self, valor):
        self.valor = valor.valor if isinstance(valor, No) else valor
        self.proximo = None
```

```python
class ListaEncadeada:
    def __init__(self):
        self.inicio = None

    # Métodos CRUD
    def criar(self, valor):
        novo = valor if isinstance(valor, No) else No(valor)
        if not self.inicio:
            self.inicio = novo
            return
        atual = self.inicio
        while atual.proximo:
            atual = atual.proximo
        atual.proximo = novo

    def ler(self):
        elementos = []
        atual = self.inicio
        while atual:
            elementos.append(atual.valor)
            atual = atual.proximo
        return elementos

    def atualizar(self, valor_antigo, valor_novo):
        atual = self.inicio
        while atual:
            if atual.valor == valor_antigo:
                atual.valor = valor_novo
                return True
            atual = atual.proximo
        return False

    def deletar(self, valor):
        atual = self.inicio
        ant = None
        while atual:
            if atual.valor == valor:
                if ant:
                    ant.proximo = atual.proximo
                else:
                    self.inicio = atual.proximo
                return True
            ant = atual
            atual = atual.proximo
        return False

    # Solução da Questão (a)
    def particionar(self, k):
        f_menor = No(0)
        f_maior = No(0)

        p_menor = f_menor
        p_maior = f_maior

        atual = self.inicio
        while atual:
            if atual.valor <= k:
                p_menor.proximo = atual
                p_menor = p_menor.proximo
            else:
                p_maior.proximo = atual
                p_maior = p_maior.proximo
            atual = atual.proximo

        p_maior.proximo = None
        p_menor.proximo = f_maior.proximo

        self.inicio = f_menor.proximo if f_menor.proximo else f_maior.proximo
        return self
```
