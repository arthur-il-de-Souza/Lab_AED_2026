#### 5. Intercalação

```python
class No:
    def __init__(self, valor):
        self.valor = valor.valor if isinstance(valor, No) else valor
        self.proximo = None
```

```python
class ListaEncadeada:
    def __init__(self):
        self.inicio = None

    # Métodos CRUD
    def criar(self, valor):
        novo = valor if isinstance(valor, No) else No(valor)
        if not self.inicio:
            self.inicio = novo
            return
        atual = self.inicio
        while atual.proximo:
            atual = atual.proximo
        atual.proximo = novo

    def ler(self):
        elementos = []
        atual = self.inicio
        while atual:
            elementos.append(atual.valor)
            atual = atual.proximo
        return elementos

    def atualizar(self, valor_antigo, valor_novo):
        atual = self.inicio
        while atual:
            if atual.valor == valor_antigo:
                atual.valor = valor_novo
                return True
            atual = atual.proximo
        return False

    def deletar(self, valor):
        atual = self.inicio
        ant = None
        while atual:
            if atual.valor == valor:
                if ant:
                    ant.proximo = atual.proximo
                else:
                    self.inicio = atual.proximo
                return True
            ant = atual
            atual = atual.proximo
        return False

    # Solução da Questão (a)
    @staticmethod
    def intercalar(l1, l2):
        lista_res = ListaEncadeada()
        fantasma = No(0)
        atual = fantasma

        p1 = l1.inicio
        p2 = l2.inicio

        while p1 and p2:
            if p1.valor <= p2.valor:
                atual.proximo = p1
                p1 = p1.proximo
            else:
                atual.proximo = p2
                p2 = p2.proximo
            atual = atual.proximo

        # Anexa o restante da lista que não chegou ao fim
        if p1:
            atual.proximo = p1
        elif p2:
            atual.proximo = p2

        lista_res.inicio = fantasma.proximo
        l1.inicio = None
        l2.inicio = None
        return lista_res
```
