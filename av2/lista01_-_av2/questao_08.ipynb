8. O elemento que aparece mais vezes

```python
class No:
    def __init__(self, valor):
        self.valor = valor.valor if isinstance(valor, No) else valor
        self.proximo = None
```

```python
class ListaEncadeada:
    def __init__(self):
        self.inicio = None

    # Métodos CRUD
    def criar(self, valor):
        novo = valor if isinstance(valor, No) else No(valor)
        if not self.inicio:
            self.inicio = novo
            return
        atual = self.inicio
        while atual.proximo:
            atual = atual.proximo
        atual.proximo = novo

    def ler(self):
        elementos = []
        atual = self.inicio
        while atual:
            elementos.append(atual.valor)
            atual = atual.proximo
        return elementos

    def atualizar(self, valor_antigo, valor_novo):
        atual = self.inicio
        while atual:
            if atual.valor == valor_antigo:
                atual.valor = valor_novo
                return True
            atual = atual.proximo
        return False

    def deletar(self, valor):
        atual = self.inicio
        ant = None
        while atual:
            if atual.valor == valor:
                if ant:
                    ant.proximo = atual.proximo
                else:
                    self.inicio = atual.proximo
                return True
            ant = atual
            atual = atual.proximo
        return False

    # Solução da Questão (a)
    def elemento_mais_frequente(self):
        if not self.inicio:
            return None, 0

        frequencias = {}
        atual = self.inicio
        elem_mais_freq = None
        max_freq = 0

        while atual:
            v = atual.valor
            frequencias[v] = frequencias.get(v, 0) + 1

            if frequencias[v] > max_freq:
                max_freq = frequencias[v]
                elem_mais_freq = v

            atual = atual.proximo

        return elem_mais_freq, max_freq
```
