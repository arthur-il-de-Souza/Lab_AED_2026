### 2. Separando pares e Ímpares

#### 2.1 Definição do Nó

```python
class No:
    def __init__(self, valor):
        self.valor = valor.valor if isinstance(valor, No) else valor
        self.proximo = None
```

#### 2.2 Definição da Lista

```python
class ListaEncadeada:
    def __init__(self):
        self.inicio = None

    # Métodos CRUD
    def criar(self, valor):
        novo = valor if isinstance(valor, No) else No(valor)
        if not self.inicio:
            self.inicio = novo
            return
        atual = self.inicio
        while atual.proximo:
            atual = atual.proximo
        atual.proximo = novo

    def ler(self):
        elementos = []
        atual = self.inicio
        while atual:
            elementos.append(atual.valor)
            atual = atual.proximo
        return elementos

    def atualizar(self, valor_antigo, valor_novo):
        atual = self.inicio
        while atual:
            if atual.valor == valor_antigo:
                atual.valor = valor_novo
                return True
            atual = atual.proximo
        return False

    def deletar(self, valor):
        atual = self.inicio
        ant = None
        while atual:
            if atual.valor == valor:
                if ant:
                    ant.proximo = atual.proximo
                else:
                    self.inicio = atual.proximo
                return True
            ant = atual
            atual = atual.proximo
        return False

    # Solução da Questão (a)
    def separar_pares_impares(self):
        l_imp = ListaEncadeada()
        l_par = ListaEncadeada()

        fim_imp = None
        fim_par = None

        atual = self.inicio
        while atual:
            prox = atual.proximo
            atual.proximo = None

            if atual.valor % 2 != 0:  # Ímpar (p1)
                if not l_imp.inicio:
                    l_imp.inicio = atual
                else:
                    fim_imp.proximo = atual
                fim_imp = atual
            else:  # Par (p2)
                if not l_par.inicio:
                    l_par.inicio = atual
                else:
                    fim_par.proximo = atual
                fim_par = atual

            atual = prox

        self.inicio = None
        return l_imp, l_par
```

#### 2.3 Separação Lista Par e Lista Impar

```python
no1 = No(11)
no2 = No(102)
no3 = No(1001)
no4 = No(10002)

lista = ListaEncadeada()

lista.criar(no1)
lista.criar(no2)
lista.criar(no3)
lista.criar(no4)

lista_imp, lista_par = lista.separar_pares_impares()

for _ in lista_imp.ler(): print(_, end=" ")
print("")
for _ in lista_par.ler(): print(_, end=" ")
```
