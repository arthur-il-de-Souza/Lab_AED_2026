### 10. Interseção

```python
class No:
  def __init__(self, valor):
    self.valor = valor.valor if isinstance(valor, No) else valor
    self.proximo = None
```

```python
class ListaEncadeada:
  def __init__(self):
    self.inicio = None

  # Métodos CRUD
  def criar(self, valor):
    novo = valor if isinstance(valor, No) else No(valor)
    if not self.inicio:
      self.inicio = novo
      return
    atual = self.inicio
    while atual.proximo:
      atual = atual.proximo
    atual.proximo = novo

  def ler(self):
    elementos = []
    atual = self.inicio
    while atual:
      elementos.append(atual.valor)
      atual = atual.proximo
    return elementos

  def atualizar(self, valor_antigo, valor_novo):
    atual = self.inicio
    while atual:
      if atual.valor == valor_antigo:
        atual.valor = valor_novo
        return True
      atual = atual.proximo
    return False

  def deletar(self, valor):
    atual = self.inicio
    ant = None
    while atual:
      if atual.valor == valor:
        if ant:
          ant.proximo = atual.proximo
        else:
          self.inicio = atual.proximo
        return True
      ant = atual
      atual = atual.proximo
    return False

  # Solução da Questão (a)
  @staticmethod
  def intersecao(l1, l2):
    elementos_l2 = set()
    atual_l2 = l2.inicio
    while atual_l2:
      elementos_l2.add(atual_l2.valor)
      atual_l2 = atual_l2.proximo

    lista_res = ListaEncadeada()
    vistos_res = set()
    fim_res = None

    atual_l1 = l1.inicio
    while atual_l1:
      val = atual_l1.valor
      if val in elementos_l2 and val not in vistos_res:
        novo = No(val)
        if not lista_res.inicio:
          lista_res.inicio = novo
        else:
          fim_res.proximo = novo
        fim_res = novo
        vistos_res.add(val)
      atual_l1 = atual_l1.proximo

    return lista_res
```
