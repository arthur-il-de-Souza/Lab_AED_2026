### 1. O maior no fim

#### 1.1 Definie Nó

```python
# Definição do Nó

class No:
    def __init__(self, valor):
        self.valor = valor
        self.proximo = None
```

#### 1.2 Definição da Lista Encadeada

```python
class ListaEncadeada:
    def __init__(self):
        self.inicio = None

    def criar(self, valor):
        novo = valor if isinstance(valor, No) else No(valor)
        if not self.inicio:
            self.inicio = novo
            return
        atual = self.inicio
        while atual.proximo:
            atual = atual.proximo
        atual.proximo = novo

    def ler(self):
        elementos = []
        atual = self.inicio
        while atual:
            elementos.append(atual.valor)
            atual = atual.proximo
        return elementos

    def atualizar(self, valor_antigo, valor_novo):
        atual = self.inicio
        while atual:
            if atual.valor == valor_antigo:
                atual.valor = valor_novo
                return True
            atual = atual.proximo
        return False

    def deletar(self, valor):
        atual = self.inicio
        ant = None
        while atual:
            if atual.valor == valor:
                if ant:
                    ant.proximo = atual.proximo
                else:
                    self.inicio = atual.proximo
                return True
            ant = atual
            atual = atual.proximo
        return False

    def mover_maior_fim(self):
        if not self.inicio or not self.inicio.proximo:
            return self

        maior = self.inicio
        ant_maior = None
        atual = self.inicio
        ant = None

        while atual:
            if atual.valor > maior.valor:
                maior = atual
                ant_maior = ant
            ant = atual
            atual = atual.proximo

        ultimo = ant

        if maior == ultimo:
            return self

        if ant_maior:
            ant_maior.proximo = maior.proximo
        else:
            self.inicio = maior.proximo

        ultimo.proximo = maior
        maior.proximo = None

        return self  # Permite encadeamento e atribuição (lista = lista.mover_maior_fim())
```

```python
no1 = No(10)
no2 = No(100)
no3 = No(1000)
no4 = No(10000)

lista = ListaEncadeada()

lista.criar(no4)
lista.criar(no1)
lista.criar(no2)
lista.criar(no3)
```

#### 1.3 Mover o Maior para o Final

```python
for _ in lista.ler(): print(_, end=" ")   
lista1 = lista.mover_maior_fim()
print("\n")
for _ in lista1.ler(): print(_, end=" ")   
```
